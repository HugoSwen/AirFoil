```python
import cadquery as cq
from jupyter_cadquery import show

r = cq.Workplane("XY").rect(0.375, 0.375).extrude(0.375)
p2 = (
    cq.Workplane("YZ", origin=(0.375, 0.125, 0))
    .rect(0.25, 0.25)
    .extrude(0.375)
    .translate((-0.375, -0.125, 0))
)
show(r.union(p2))
```
