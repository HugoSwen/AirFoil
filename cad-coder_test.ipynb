````python
import os
import openai
from cad_coder_prompt import *
import re
from dotenv import load_dotenv
load_dotenv()

if os.getenv("OPENAI_API_KEY"):
    client = openai.OpenAI(
        api_key=os.getenv("OPENAI_API_KEY"),
        base_url='https://api.openai-proxy.org/v1',
    )
elif os.getenv("SILICONFLOW_API_KEY"):
    client = openai.OpenAI(
        api_key=os.getenv("SILICONFLOW_API_KEY"),
        base_url='https://api.siliconflow.cn/v1',
    )
````

````python
def llm_query(prompt: str, model: str = "Qwen/Qwen2.5-72B-Instruct", temperature: float = 0.8):
    """
    向LLM发送查询并返回响应
    """
    for _ in range(3):
        try:
            if model == "Qwen/Qwen3-32B" or model == "Qwen/Qwen3-14B" or model == "Qwen/Qwen3-8B" or model == "zai-org/GLM-4.6" or model == "Qwen/Qwen3-235B-A22B" or model == "deepseek-ai/DeepSeek-V3.1":
                response = client.chat.completions.create(
                    model=model,
                    messages=[
                        {"role": "user", "content": prompt}
                    ],
                    temperature=temperature,
                    extra_body={"enable_thinking": False}
                )
            else:
                response = client.chat.completions.create(
                    model=model,
                    messages=[
                        {"role": "user", "content": prompt}
                    ],
                    temperature=temperature
                )

            print(f"{response.choices[0].message.content}")

            return response.choices[0].message.content
        except Exception as e:
            print(f"LLM query failed: {e}, retrying...")
            continue
````

````python
def execute_extracted_code(response: str):   
    # 提取python代码并执行
    pattern = r"```python(.*?)```"
    match = re.search(pattern, response, re.DOTALL)
    if match:
        print("Executing code...")
    #     code = match.group(1).strip()
    #     exec(code)
    
    # else:
    #     print("No executable code found in the response.")
````
